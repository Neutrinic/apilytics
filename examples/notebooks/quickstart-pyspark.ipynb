# Apilytics Quickstart (PySpark)

Query REST APIs with Spark SQL using Apilytics.

## Setup

When running via `docker compose`, two catalogs are pre-configured:
- `pokeapi` - Query PokeAPI (pokemon, types, abilities)
- `github` - Query GitHub API (issues from octocat/Hello-World)

The SparkSession is available as `spark`.

```python
# SparkSession is pre-configured when running via docker/compose
# Two catalogs available: pokeapi and github
print(f"Spark version: {spark.version}")
print(f"PokeAPI catalog: {spark.conf.get('spark.sql.catalog.pokeapi', 'NOT SET')}")
print(f"GitHub catalog: {spark.conf.get('spark.sql.catalog.github', 'NOT SET')}")
```

## Basic Queries

Query the PokeAPI like a database table:

```python
# List available tables in PokeAPI catalog
spark.sql("SHOW TABLES IN pokeapi.default").show()
```

```python
# Query Pokemon
spark.sql("SELECT name, url FROM pokeapi.default.pokemon LIMIT 10").show(truncate=False)
```

```python
# Check the schema
spark.sql("DESCRIBE pokeapi.default.pokemon").show()
```

## DataFrame API

You can also use the DataFrame API:

```python
# Load as DataFrame
pokemon = spark.table("pokeapi.default.pokemon")

# Filter and select
pokemon.filter(pokemon.name.startswith("char")).show()
```

```python
# Count
print(f"Total Pokemon: {pokemon.count()}")
```

## Caching

Each query hits the API. For repeated analysis, cache locally:

```python
# Cache for faster repeated queries
pokemon_cached = spark.table("pokeapi.default.pokemon").cache()

# Now these queries are fast
pokemon_cached.groupBy("name").count().show(5)
pokemon_cached.filter(pokemon_cached.name.like("%saur")).show()
```

## Query GitHub API

The `github` catalog is also pre-configured:

```python
# Query GitHub issues
spark.sql("SELECT number, title, state FROM github.default.issues LIMIT 5").show(truncate=40)
```
