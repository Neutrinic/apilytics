# Apilytics Quickstart (PySpark)

Query REST APIs with Spark SQL using Apilytics.

## Setup

When running via `docker compose` or `docker run ... jupyter`, the catalog is pre-configured.
The SparkSession is available as `spark`.

```python
# SparkSession is pre-configured when running via docker/compose
# Just verify it's available
print(f"Spark version: {spark.version}")
print(f"Catalog configured: {spark.conf.get('spark.sql.catalog.api', 'NOT SET')}")
```

## Basic Queries

Query the PokeAPI like a database table:

```python
# List available tables
spark.sql("SHOW TABLES IN api.default").show()
```

```python
# Query Pokemon
spark.sql("SELECT name, url FROM api.default.pokemon LIMIT 10").show(truncate=False)
```

```python
# Check the schema
spark.sql("DESCRIBE api.default.pokemon").show()
```

## DataFrame API

You can also use the DataFrame API:

```python
# Load as DataFrame
pokemon = spark.table("api.default.pokemon")

# Filter and select
pokemon.filter(pokemon.name.startswith("char")).show()
```

```python
# Count
print(f"Total Pokemon: {pokemon.count()}")
```

## Caching

Each query hits the API. For repeated analysis, cache locally:

```python
# Cache for faster repeated queries
pokemon_cached = spark.table("api.default.pokemon").cache()

# Now these queries are fast
pokemon_cached.groupBy("name").count().show(5)
pokemon_cached.filter(pokemon_cached.name.like("%saur")).show()
```
