# PokeAPI Data Exploration

Explore the Pokemon universe with SQL queries using Apilytics.

The SparkSession (`spark`) is pre-configured with the PokeAPI catalog.

```python
# Verify setup
print(f"Spark version: {spark.version}")
```

## Discover Available Tables

```python
spark.sql("SHOW TABLES IN api.default").show()
```

## Pokemon Overview

```python
# Get first 20 Pokemon
pokemon = spark.sql("""
    SELECT name, url 
    FROM api.default.pokemon 
    LIMIT 20
""")
pokemon.show(truncate=False)
```

```python
# Count total Pokemon
total = spark.sql("SELECT COUNT(*) as total FROM api.default.pokemon").collect()[0][0]
print(f"Total Pokemon in the API: {total}")
```

## Pokemon Types

```python
# List all Pokemon types
types = spark.sql("SELECT name FROM api.default.types")
types.show()
```

## Analyze Pokemon Names

```python
# Cache Pokemon data for faster analysis
pokemon_df = spark.table("api.default.pokemon").cache()
print(f"Cached {pokemon_df.count()} Pokemon")
```

```python
# Find Pokemon by name pattern
pokemon_df.filter(pokemon_df.name.like("%chu%")).show()
```

```python
# Pokemon names starting with each letter
spark.sql("""
    SELECT 
        UPPER(SUBSTRING(name, 1, 1)) as first_letter,
        COUNT(*) as count
    FROM api.default.pokemon
    GROUP BY UPPER(SUBSTRING(name, 1, 1))
    ORDER BY count DESC
    LIMIT 10
""").show()
```

```python
# Longest Pokemon names
spark.sql("""
    SELECT name, LENGTH(name) as name_length
    FROM api.default.pokemon
    ORDER BY name_length DESC
    LIMIT 10
""").show()
```

## Abilities

```python
# List abilities
spark.sql("SELECT name FROM api.default.abilities LIMIT 20").show()
```
